## Step 1: Feature Engineering

```python
import os
from dotenv import load_dotenv

load_dotenv()

SILVER_CLEANED_DATA_PATH = os.getenv(
    "SILVER_CLEANED_DATA_PATH",
    os.path.join("..", "data", "silver", "Silver_Cleaned_Netflix_Data.csv")
)
print("File Loaded!")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(SILVER_CLEANED_DATA_PATH)
print("First 5 rows of the dataset:")
df.head()
```

```python
print("Last 5 rows of the dataset:")
df.tail()
```

```python
print("Size of the DataFrame:", df.size)
print("Shape of the DataFrame:", df.shape)
print("\nColumns in the DataFrame:", df.columns)
```

```python
print("The number of Null Values : ", df.isnull().sum().sum())
```

```python
# Work on a copy to keep raw data intact
df_fe = df.copy()
df_fe["date_added"] = pd.to_datetime(df_fe["date_added"], errors="coerce")
df_fe.head()
```

```python
df_fe["is_movie"] = (df_fe["type"].str.strip().str.lower() == "movie").astype(int)
df_fe["is_movie"].value_counts()
```

```python
# Extract and clean primary country
df_fe["primary_country"] = df_fe["country"].str.split(",").str[0].str.strip()
df_fe["primary_genre"] = df_fe["listed_in"].str.split(",").str[0].str.strip()

print("\nSample primary_country values:")
df_fe[["country", "primary_country"]].head()
```

```python
print("\nSample primary_genre values:")
df_fe[["listed_in", "primary_genre"]].head() 
```

### Core Feature Creation
Create business features used for downstream modeling: content type flags, primary category fields, content age, and text/count-based signals.

```python
df_fe["duration_num"] = (
    df_fe["duration"].astype(str)
    .str.extract(r"(\d+)", expand=False)
    .astype(float)
)

df_fe["num_countries"] = df_fe["country"].fillna("").str.split(",").apply(
    lambda x: len([v for v in x if v.strip()])
)

df_fe["num_genres"] = df_fe["listed_in"].fillna("").str.split(",").apply(
    lambda x: len([v for v in x if v.strip()])
)

df_fe["cast_count"] = df_fe["cast"].fillna("").str.split(",").apply(
    lambda x: len([v for v in x if v.strip()])
)

df_fe["title_word_count"] = df_fe["title"].fillna("").str.split().str.len()

df_fe["description_word_count"] = (
    df_fe["description"].fillna("").str.split().str.len()
)
```

```python
if "year_added" not in df_fe.columns:
    df_fe["year_added"] = df_fe["date_added"].dt.year

if "month_added" not in df_fe.columns:
    df_fe["month_added"] = df_fe["date_added"].dt.month

if "day_added" not in df_fe.columns:
    df_fe["day_added"] = df_fe["date_added"].dt.day

# check result
print("\nSample date_added features:")
df_fe[["date_added", "year_added", "month_added", "day_added"]].head()
```

```python
df_fe["content_age"] = df_fe["year_added"] - df_fe["release_year"]
df_fe["content_age_missing"] = df_fe["content_age"].isna().astype(int)

# Keep previous feature name for compatibility
df_fe["content_age_when_added"] = df_fe["content_age"]

df_fe["release_decade"] = (df_fe["release_year"] // 10) * 10

# check result
print("\nSample content age features:")
df_fe[
    ["release_year", 
    "year_added", 
    "content_age", 
    "content_age_missing", 
    "content_age_when_added", 
    "release_decade"]
].head()
```

```python
numeric_features = [
    "release_year",
    "release_decade",
    "year_added",
    "month_added",
    "day_added",
    "content_age",
    "content_age_missing",
    "content_age_when_added",
    "date_added_missing",
    "duration_num",
    "num_countries",
    "num_genres",
    "cast_count",
    "title_word_count",
    "description_word_count",
    "is_movie",
]

# Keep only available columns
numeric_features = [c for c in numeric_features if c in df_fe.columns]
```

### Feature Scaling
Prepare numeric feature list, handle missing numeric values, and apply `StandardScaler` to normalize ranges for machine learning.

```python
df_fe[numeric_features] = df_fe[numeric_features].apply(lambda col: col.fillna(col.median()))
df_fe.head()
```

```python
# Feature Scaling
scaler = StandardScaler()

scaled_values = scaler.fit_transform(df_fe[numeric_features])

scaled_col_names = [f"{col}_scaled" for col in numeric_features]

df_scaled = pd.DataFrame(
    scaled_values,
    columns=scaled_col_names,
    index=df_fe.index
)
df_scaled.head()
```

```python
df_model = pd.concat([df_fe, df_scaled], axis=1)

print("Feature engineering + scaling complete")
print("Rows, columns:", df_model.shape)
print("Scaled columns:", len(scaled_col_names))

df_model[scaled_col_names].head()
```

### Final Dataset and Export
Combine engineered + scaled features, run quick quality checks, and save the final dataset to the gold layer.

```python
feature_cols = [
    "is_movie",
    "primary_country",
    "primary_genre",
    "content_age",
    "content_age_missing",
    "duration_num",
    "release_decade",
]
available_feature_cols = [c for c in feature_cols if c in df_fe.columns]

print("Engineered features:", available_feature_cols)
print("\nNulls in engineered feature columns:")
df_fe[available_feature_cols].isnull().sum()
```

```python
output_dir = os.path.join("..", "data", "gold")
os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(output_dir, "Netflix_Featured_Scaled_Data.csv")
df_model.to_csv(output_path, index=False)

print("Saved file!")
```
