## Step 2: Data Cleaning

```python
# Standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print("Standardized column names:", df.columns)
```

```python
# Check Missing Values
df.isnull().sum().sort_values(ascending=False)
```

```python
# Handle Missing Values
df.fillna({
    'director': 'Unknown',
    'cast': 'Unknown',
    'country': 'Unknown',
    'rating': 'Not Rated',
    'duration': '0 min',
    'listed_in': 'Unknown',
    'description': 'No Description'
}, inplace=True)
```

```python
df[['director', 'cast', 'country', 'rating', 'duration', 'listed_in', 'description']].isnull()
```

```python
df.isnull().sum()
```

```python
# 1) Keep original date_added safe
# 2) Clean in a new column
# 3) Keep missing indicator
# 4) Impute with mid-year of release_year
# 5) Create business features

df["date_added_cleaned"] = pd.to_datetime(df["date_added"], errors="coerce")

df["date_added_missing"] = df["date_added_cleaned"].isna().astype(int)

release_year_date = pd.to_datetime(
    df["release_year"].astype(str) + "-07-01", errors="coerce"
)
df["date_added_cleaned"] = df["date_added_cleaned"].fillna(release_year_date)

# Keep the original column null-free for downstream checks
df["date_added"] = df["date_added_cleaned"]

df["year_added"] = df["date_added_cleaned"].dt.year
df["month_added"] = df["date_added_cleaned"].dt.month
df["day_added"] = df["date_added_cleaned"].dt.day

df["content_age_when_added"] = df["year_added"] - df["release_year"]

print("Missing in date_added_cleaned after imputation:", df["date_added_cleaned"].isna().sum())
print("Missing in date_added after replacement:", df["date_added"].isna().sum())
print("Rows originally missing date_added:", df["date_added_missing"].sum())

df[[
    "date_added",
    "date_added_cleaned",
    "date_added_missing",
    "release_year",
    "year_added",
    "month_added",
    "day_added",
    "content_age_when_added",
]].head()
```

```python
# Final cleaning quality checks
text_cols = df.select_dtypes(include=["object", "string"]).columns
df[text_cols] = df[text_cols].apply(lambda col: col.str.strip())

duplicates_before = df.duplicated().sum()
df.drop_duplicates(inplace=True)
duplicates_after = df.duplicated().sum()

print("Total null values:", int(df.isnull().sum().sum()))
print("Duplicates before:", int(duplicates_before))
print("Duplicates after:", int(duplicates_after))
print("Current shape:", df.shape)

# Quick check
df[["date_added_missing", "year_added", "month_added", "day_added", "content_age_when_added"]].describe()
```
