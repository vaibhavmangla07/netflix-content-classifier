# Working with SQLite3

```python
import os
import sqlite3
import pandas as pd
```

```python
# Create and connect to SQLite database
db_dir = os.path.join("..", "database")
os.makedirs(db_dir, exist_ok=True)
db_path = os.path.join(db_dir, "netflix.db")

conn = sqlite3.connect(db_path)
print("Database connected successfully!")
```

```python
# Load data into SQLite table (create/replace)
gold_path = os.path.join("..", "data", "gold", "Netflix_Featured_Scaled_Data.csv")
df = pd.read_csv(gold_path)
df.to_sql("netflix_data", conn, if_exists="replace", index=False)

# Verify row count
query = "SELECT COUNT(*) AS total_rows FROM netflix_data"
pd.read_sql(query, conn)
```

```python
# Total number of titles
pd.read_sql(
    "SELECT COUNT(*) AS total_titles FROM netflix_data",
    conn
)
```

```python
# Movies vs TV Shows count
pd.read_sql(
    """
    SELECT type, COUNT(*) AS count
    FROM netflix_data
    GROUP BY type
    """,
    conn
)
```

```python
# Top 10 countries by content count
pd.read_sql(
    """
    SELECT primary_country, COUNT(*) AS count
    FROM netflix_data
    GROUP BY primary_country
    ORDER BY count DESC
    LIMIT 10
    """,
    conn
)
```

```python
# Content added per year
pd.read_sql(
    """
    SELECT year_added, COUNT(*) AS count
    FROM netflix_data
    GROUP BY year_added
    ORDER BY year_added
    """,
    conn
)
```

```python
conn.close()
print("Database connection closed")
```
